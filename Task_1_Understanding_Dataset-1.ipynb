# Executive Summary

## Customer Churn Analysis – Telecommunications Industry

### Project Overview

This project investigates customer churn within a telecommunications company using a dataset containing 7,043 customer records and 21 columns. The primary objective was to identify customer characteristics and service-related patterns associated with customer churn and translate the findings into practical business recommendations.

### Analytical Approach

The analysis was conducted using Python in Jupyter Notebook, with Pandas for data manipulation and analysis and data visualisation techniques to communicate important patterns.

The project involved:

- Dataset inspection and understanding.
- Data quality investigation and cleaning.
- Exploratory Data Analysis (EDA).
- Customer segmentation based on tenure and monthly charges.
- Comparative analysis of customer characteristics and churn.
- Advanced analysis combining contract type, payment method and monthly charge segments.
- Development of evidence-based business recommendations.

### Major Findings

The analysis identified several important patterns:

1. Month-to-month customers recorded a churn rate of 42.71%, compared with 2.83% among two-year contract customers.
2. Customers with 0–12 months of tenure recorded a churn rate of 47.44%, making early-tenure customers an important retention priority.
3. Electronic check users recorded a churn rate of 45.29%, considerably higher than users of automatic payment methods.
4. Customers in the high monthly charge segment recorded a churn rate of 34.09%, compared with 15.87% among low-charge customers.
5. High-charge month-to-month customers recorded a churn rate of 52.61%.
6. Senior citizens recorded a churn rate of 41.68%, compared with 23.61% among non-senior customers.

### Business Implications

The findings suggest that customer retention strategies should pay particular attention to early-tenure customers, month-to-month subscribers, electronic check users and customers paying higher monthly charges.

The project proposes targeted retention programmes, improved early customer engagement, payment-experience investigations, pricing reviews and customer-support improvements.

### Overall Outcome

The project demonstrates how data analysis can be used to identify customer retention patterns, investigate potential business risks and support evidence-based decision-making.

The findings and recommendations provide a foundation for further investigation and the development of targeted customer retention strategies.

# Task 2: Data Cleaning

## 2.1 Handling Missing and Blank Values

This section focuses on improving the quality and consistency of the customer churn dataset.

The initial investigation identified 11 whitespace-only entries in the TotalCharges column. These entries will be converted into proper missing values before numerical conversion.

All customer records will be preserved during the cleaning process.

```python
# ============================================
# TASK 2: DATA CLEANING: Create a Working Copy
# ============================================

# Preserve the original dataset and create a working copy
df_clean = df.copy()

# Confirm that the working copy was created
print("Working copy created successfully!")

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
# Import NumPy for numerical operations and missing-value handling
import numpy as np
```

```python
# ==========================================
# TASK 2: HANDLE MISSING AND BLANK VALUES : Convert Blank Spaces to Missing Values
# =========================================

# Replace whitespace-only entries in TotalCharges with missing values
df_clean["TotalCharges"] = df_clean["TotalCharges"].replace(
    r"^\s*$",
    np.nan,
    regex=True
)

# Count missing values after replacement
print("Missing values after cleaning:")
print(df_clean["TotalCharges"].isnull().sum())
```

```python
# ==========================================
# TASK 2: DATA TYPE CORRECTION: 2.3 Convert TotalCharges to Numeric
# ==========================================

# Convert TotalCharges from text to numeric values
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

# Display the updated data type
print("Updated TotalCharges data type:")
print(df_clean["TotalCharges"].dtype)

# Confirm the number of missing values
print("\nMissing TotalCharges values:")
print(df_clean["TotalCharges"].isnull().sum())
```

```python
# ==========================================
# TASK 2: DATA QUALITY VALIDATION: 2.4 Verify Updated Dataset Structure
# ==========================================

# Display updated data types for all columns
print("Updated dataset data types:")
print(df_clean.dtypes)

# Check the dimensions of the cleaned dataset
print("\nCleaned dataset dimensions:")
print(df_clean.shape)

# Count missing values across all columns
print("\nMissing values by column:")
print(df_clean.isnull().sum())
```

```python
# ==========================================
# TASK 2: DATA QUALITY VALIDATION: 2.4 Summary of Dataset Quality
# ==========================================

# Create a summary of column data types and missing values
data_quality_summary = pd.DataFrame({
    "Data Type": df_clean.dtypes,
    "Missing Values": df_clean.isnull().sum()
})

# Display the data quality summary
print("DATA QUALITY SUMMARY")
display(data_quality_summary)

# Display overall dataset dimensions
print("\nCleaned dataset dimensions:", df_clean.shape)

# Display total missing values
print("Total missing values:", df_clean.isnull().sum().sum())
```

### 4.5 Clustered Comparison of Average Monthly Charges by Tenure Group and Churn Status

```python
# Set the positions and width of the bars
x = np.arange(len(monthly_charges_by_tenure_churn.index))
bar_width = 0.35

# Create the figure
plt.figure(figsize=(11, 7))

# Plot retained customers
bars_no = plt.bar(
    x - bar_width / 2,
    monthly_charges_by_tenure_churn["No"],
    bar_width,
    label="Retained (No)"
)

# Plot churned customers
bars_yes = plt.bar(
    x + bar_width / 2,
    monthly_charges_by_tenure_churn["Yes"],
    bar_width,
    label="Churned (Yes)"
)

# Add average charge labels above each bar
for bars in [bars_no, bars_yes]:
    for bar in bars:
        height = bar.get_height()

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height + 1,
            f"{height:.2f}",
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold"
        )

# Annotate the difference between the two bars in each group
for i in range(len(x)):
    retained = monthly_charges_by_tenure_churn["No"].iloc[i]
    churned = monthly_charges_by_tenure_churn["Yes"].iloc[i]
    difference = churned - retained

    plt.text(
        x[i],
        churned + 7,
        f"Gap: {difference:.2f}",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

# Add chart title and axis labels
plt.title(
    "Average Monthly Charges by Tenure Group and Churn Status",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Customer Tenure Group", fontsize=12)
plt.ylabel("Average Monthly Charges", fontsize=12)

# Configure the x-axis and y-axis
plt.xticks(x, monthly_charges_by_tenure_churn.index)
plt.ylim(0, 115)

# Add legend and grid
plt.legend(title="Customer Status")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### 5.5 Churn Rate Analysis by Monthly Charge Segments

This section examines how customer churn varies across different monthly charge levels. Customers will be grouped into three charge segments to compare their distribution and churn rates.

The objective is to determine whether observed churn rates differ across monthly charge segments and identify any patterns that may be relevant to customer retention analysis.

```python
# Divide customers into three monthly charge segments
df_clean["monthly_charge_segment"] = pd.qcut(
    df_clean["monthly_charges"],
    q=3,
    labels=["Low", "Medium", "High"]
)

# Display the number of customers in each segment
display(
    df_clean["monthly_charge_segment"].value_counts().sort_index()
)
```

```python
# Calculate customer count and churn rate by monthly charge segment
monthly_charge_churn = (
    df_clean.groupby("monthly_charge_segment", observed=False)
    .agg(
        customer_count=("customer_id", "count"),
        churned_customers=(
            "churn",
            lambda x: (x == "Yes").sum()
        ),
        churn_rate=(
            "churn",
            lambda x: (x == "Yes").mean() * 100
        )
    )
)

# Round the results for readability
monthly_charge_churn = monthly_charge_churn.round(2)

# Display the summary
display(monthly_charge_churn)
```

### 5.5.1 Interpretation of Churn Rates by Monthly Charge Segment

The analysis reveals differences in customer churn rates across the three monthly charge segments.

The Low segment recorded the lowest churn rate at **15.87%**, followed by the Medium segment at **29.68%**. The High segment recorded the highest churn rate at **34.09%**.

The difference between the High and Low segments is **18.22 percentage points**, indicating a higher observed churn rate among customers in the High monthly charge segment.

**Business Insight:**

Customers in higher monthly charge segments demonstrate higher observed churn rates in this dataset. This finding suggests that customers with higher recurring charges may deserve further attention in customer retention analysis.

**Analytical Limitation:**

This is an observed association and does not establish that higher monthly charges cause churn. Contract type, tenure and other customer characteristics may also contribute to the differences. Further investigation would be necessary before making causal conclusions or recommending pricing changes.
